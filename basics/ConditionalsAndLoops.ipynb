IF ELSE BLOCK

```python
x = int(input("Enter a number"))
if x==5:
  print(x)
  print("If block")
else:
  print(x)
  print("Else block")
```

ELIF

```python
x = int(input("Enter a number"))
if x<5:
  print(x)
  print("If block")
elif x<3:
  print(x)
  print("Elif block")
else:
  print(x)
  print("Else block")
```

NESTED IF

```python
x = int(input("Enter a number"))
if x>5:
  print("IF")
  if x<7:
    print("Nested if")
    print("x is between 5 and 7")
  else:
    print("Nested else")
print("Outside")
```

Break  & Cotinue

```python
n=0
while n<5:
  print(n)
  n = n+1
print("Outside")
```

```python
n=0
while n<5:
  if n==2:
    print("Inside If")
    continue
  print(n)
  n = n+1
print("Outside")
```

```python
n=0
while n<5:
  if n==2:
    print("Inside If")
    break
  print(n)
  n = n+1
print("Outside")
```
