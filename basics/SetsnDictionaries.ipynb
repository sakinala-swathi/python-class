Dictionaries- Store data in key-value pair.
4 major categories.
 1. Ordered.
2.  Indexed with the key-value pair type.
  3. Mutable
  4. Duplication is not allowed

```python
student= ["Swathi", 18, "Female"]
```

```python
student[0]
```

```python
student2 = {
    "name": "Swathi",
    "age": 18,
    "gender": "Female"
}
```
